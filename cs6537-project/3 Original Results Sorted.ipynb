```python
import pandas as pd

# ================= 1. 修改这里的文件名 =================
file_name = r"output\DeepSeek Results.csv" 
output_name = r"output\DeepSeek Results_Sorted.csv"

# ================= 2. 执行排序 =================
print(f"正在处理: {file_name}")

# 读取 CSV
df = pd.read_csv(file_name, encoding='utf-8-sig')

# 核心：确保 Run_ID 是数字类型，否则会出现 1, 10, 100, 2 这种错误的排序
df['Run_ID'] = pd.to_numeric(df['Run_ID'], errors='coerce')

# 设定排序优先级：品类 -> 语言 -> 格式 -> 提问模板 -> 实验ID
sort_columns = ['Category', 'Language', 'Format', 'Guide_Type', 'Run_ID']

# 执行升序排列
df_sorted = df.sort_values(by=sort_columns, ascending=True)

# 保存文件
df_sorted.to_csv(output_name, index=False, encoding='utf-8-sig')

print(f"✅ 排序完成！已保存为: {output_name}")
```
